```python
import pandas as pd
import numpy as np

# Import CSV file
try:
    df = pd.read_excel("Logic_Gates_With_Step_Activation.xlsx")
except:
    df = pd.read_excel("Lab7/Logic_Gates_With_Step_Activation.xlsx")

# Take only the header
header = df.columns

print("Header:")
print(header.tolist())

# XOR Gate
epochs = int(input("Enter number of epochs: "))
learning_rate = float(input("Enter learning rate: "))
w1 = float(input("Enter initial weight w1: "))
w2 = float(input("Enter initial weight w2: "))
b = float(input("Enter initial bias: "))

X = np.array([[0,0],[0,1],[1,0],[1,1]])

Y = np.array([0, 1, 1, 0])
iteration = 1
# Print header
print("\nEpoch\tIter\tx1\tx2\tTarget(y)\tw1_in\tw2_in\tb_in\tz\tPrediction\tError(e)\tw1_out\tw2_out\tb_out")

for epoch in range(1, epochs + 1):

    for i in range(len(X)):

        x1 = X[i, 0]
        x2 = X[i, 1]
        target = Y[i]

        # Input weights
        w1_in = w1
        w2_in = w2
        b_in = b

        # Calculate z
        z = w1_in * x1 + w2_in * x2 + b_in

        # Step activation
        if z >= 0:
            prediction = 1
        else:
            prediction = 0

        # Error
        error = target - prediction

        # Weight update
        w1_out = w1_in + learning_rate * error * x1
        w2_out = w2_in + learning_rate * error * x2
        b_out = b_in + learning_rate * error

        # Print row
        print(
            f"{epoch}\t{iteration}\t{x1}\t{x2}\t{target}\t"
            f"{w1_in:.1f}\t{w2_in:.1f}\t{b_in:.1f}\t{z:.1f}\t"
            f"{prediction}\t{error}\t{w1_out:.1f}\t"
            f"{w2_out:.1f}\t{b_out:.1f}"
        )

        # Update weights
        w1 = w1_out
        w2 = w2_out
        b = b_out

        iteration += 1
```

```python
import pandas as pd
import numpy as np

try:
    df = pd.read_excel("XOR_MLP_With_Hyperparameters_No_BPG.xlsx", skiprows=16)
except:
    df = pd.read_excel("Exercise/XOR_MLP_With_Hyperparameters_No_BPG.xlsx", skiprows=16)

print("Header:")
print(df.columns.tolist())

print("\nReference Table from Excel:")
print(df.to_string(index=False))
choice = input("\nDo you want to enter custom weights and biases? (y/n): ")

if choice.lower() == 'y':
    w11 = float(input("Enter w11 (x1 -> h1): "))
    w21 = float(input("Enter w21 (x2 -> h1): "))
    b_h1 = float(input("Enter b_h1 (bias h1): "))

    w12 = float(input("Enter w12 (x1 -> h2): "))
    w22 = float(input("Enter w22 (x2 -> h2): "))
    b_h2 = float(input("Enter b_h2 (bias h2): "))

    v1 = float(input("Enter v1 (h1 -> y): "))
    v2 = float(input("Enter v2 (h2 -> y): "))
    b_y = float(input("Enter b_y (bias y): "))
else:
    w11 = 1.0
    w21 = 1.0
    b_h1 = -0.5

    w12 = -1.0
    w22 = -1.0
    b_h2 = 1.5

    v1 = 1.0
    v2 = 1.0
    b_y = -1.5

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([0, 1, 1, 0])

epochs = 1
iteration = 1

# Print header
print("\nEpoch\tIter\tx1\tx2\tTarget\tz_h1\th1\tz_h2\th2\tz_y\tPred\tError")

for epoch in range(1, epochs + 1):

    for i in range(len(X)):

        x1 = X[i, 0]
        x2 = X[i, 1]
        target = Y[i]

        # Hidden neuron 1 (OR logic)
        z_h1 = w11 * x1 + w21 * x2 + b_h1
        if z_h1 >= 0:
            h1 = 1
        else:
            h1 = 0

        # Hidden neuron 2 (NAND logic)
        z_h2 = w12 * x1 + w22 * x2 + b_h2
        if z_h2 >= 0:
            h2 = 1
        else:
            h2 = 0

        # Output neuron (AND logic)
        z_y = v1 * h1 + v2 * h2 + b_y
        if z_y >= 0:
            prediction = 1
        else:
            prediction = 0

        # Error
        error = target - prediction

        # Print row
        print(
            f"{epoch}\t{iteration}\t{x1}\t{x2}\t{target}\t"
            f"{z_h1:.1f}\t{h1}\t{z_h2:.1f}\t{h2}\t"
            f"{z_y:.1f}\t{prediction}\t{error}"
        )

        iteration += 1

print("\nForward pass complete! All errors = 0.")
```

```python
import pandas as pd
import numpy as np

try:
    df = pd.read_excel("XOR_MLP_Backprop2.xlsx", skiprows=16)
except:
    df = pd.read_excel("Exercise/XOR_MLP_Backprop2.xlsx", skiprows=16)

# Print Reference Header & First Epoch from Excel
print("Header:")
print(df.columns.tolist())

print("\nReference (Epoch 1) from Excel:")
print(df.head(4)[['Epoch', 'Iter', 'x1', 'x2', 'Target', 'h1', 'h2', 'y_hat', 'Error', 'MSE', 'Correct?']].to_string(index=False))


def sigmoid(z):
    return 1 / (1 + np.exp(-z))

epochs_input = input("\nEnter number of epochs (default 50): ")
epochs = int(epochs_input) if epochs_input.strip() else 50

lr_input = input("Enter learning rate (default 2.5): ")
learning_rate = float(lr_input) if lr_input.strip() else 2.5

choice = input("Do you want to enter custom initial weights and biases? (y/n): ")

if choice.lower() == 'y':
    w11 = float(input("Enter w11 (x1 -> h1): "))
    w21 = float(input("Enter w21 (x2 -> h1): "))
    b1 = float(input("Enter b1 (bias h1): "))

    w12 = float(input("Enter w12 (x1 -> h2): "))
    w22 = float(input("Enter w22 (x2 -> h2): "))
    b2 = float(input("Enter b2 (bias h2): "))

    w1 = float(input("Enter w1 (h1 -> y): "))
    w2 = float(input("Enter w2 (h2 -> y): "))
    b3 = float(input("Enter b3 (bias y): "))
else:
    # Initial weights & biases from Excel
    w11 = 2.5
    w21 = 2.5
    b1 = -1.0

    w12 = -2.5
    w22 = -2.5
    b2 = 3.5

    w1 = 3.0
    w2 = 3.0
    b3 = -4.5

X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

Y = np.array([0, 1, 1, 0])

iteration = 1

# Print header
print("\nEpoch\tIter\tx1\tx2\tTarget\th1\th2\tz_y\ty_hat\tPred\tError\tMSE\tStatus")

for epoch in range(1, epochs + 1):

    for i in range(len(X)):

        x1 = X[i, 0]
        x2 = X[i, 1]
        target = Y[i]

        # --------------------------------
        # 1. Forward Pass
        # --------------------------------

        # Hidden Neuron 1
        z_h1 = w11 * x1 + w21 * x2 + b1
        h1 = sigmoid(z_h1)

        # Hidden Neuron 2
        z_h2 = w12 * x1 + w22 * x2 + b2
        h2 = sigmoid(z_h2)

        # Output Neuron
        z_y = w1 * h1 + w2 * h2 + b3
        y_hat = sigmoid(z_y)

        # Classification and Error
        if y_hat >= 0.5:
            y_classified = 1
        else:
            y_classified = 0

        error = target - y_hat
        mse = 0.5 * (error ** 2)

        if y_classified == target:
            status = "PASS"
        else:
            status = "FAIL"

        delta_y = error * y_hat * (1 - y_hat)
        delta_h1 = delta_y * w1 * h1 * (1 - h1)
        delta_h2 = delta_y * w2 * h2 * (1 - h2)

        # Output layer weights update
        w1_out = w1 + learning_rate * delta_y * h1
        w2_out = w2 + learning_rate * delta_y * h2
        b3_out = b3 + learning_rate * delta_y

        # Hidden layer weights update
        w11_out = w11 + learning_rate * delta_h1 * x1
        w21_out = w21 + learning_rate * delta_h1 * x2
        b1_out = b1 + learning_rate * delta_h1

        w12_out = w12 + learning_rate * delta_h2 * x1
        w22_out = w22 + learning_rate * delta_h2 * x2
        b2_out = b2 + learning_rate * delta_h2

        # Print row
        print(
            f"{epoch}\t{iteration}\t{x1}\t{x2}\t{target}\t"
            f"{h1:.4f}\t{h2:.4f}\t{z_y:.4f}\t{y_hat:.4f}\t"
            f"{y_classified}\t{error:.4f}\t{mse:.4f}\t{status}"
        )

        # Update weights for next iteration
        w1 = w1_out
        w2 = w2_out
        b3 = b3_out

        w11 = w11_out
        w21 = w21_out
        b1 = b1_out

        w12 = w12_out
        w22 = w22_out
        b2 = b2_out

        iteration += 1

print("\n--- Final Trained Weights ---")
print(f"w11 = {w11:.4f}, w21 = {w21:.4f}, b1 = {b1:.4f}")
print(f"w12 = {w12:.4f}, w22 = {w22:.4f}, b2 = {b2:.4f}")
print(f"w1  = {w1:.4f}, w2  = {w2:.4f}, b3 = {b3:.4f}")

print("\n--- Final Testing Evaluation ---")
print("x1\tx2\tTarget\ty_hat\tPrediction\tResult")
for i in range(len(X)):
    x1 = X[i, 0]
    x2 = X[i, 1]
    target = Y[i]
    h1 = sigmoid(w11 * x1 + w21 * x2 + b1)
    h2 = sigmoid(w12 * x1 + w22 * x2 + b2)
    y_hat = sigmoid(w1 * h1 + w2 * h2 + b3)
    pred = 1 if y_hat >= 0.5 else 0
    res = "PASS" if pred == target else "FAIL"
    print(f"{x1}\t{x2}\t{target}\t{y_hat:.4f}\t{pred}\t\t{res}")
```
